```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
```

## 5.1 Introduction to pandas Data Structures

### Series

A Series is a one-dimensional array-like object containing a sequence of values of the same type and an associated array of data labels, called its index.

```python
obj = pd.Series([4, 5, -7, 3])
```

```python
obj
```

```python
obj2 = pd.Series([1, 2, 3, 4], index=['a', 'b', 'c', 'd'])
```

```python
obj2
```

```python
obj2.index
```

```python
obj2['a']
```

```python
obj2['b'] = 6
```

```python
obj2[['a', 'b']] = -1
```

```python
obj2
```

Using Numpy functions or Numpy-like operations, such as filtering with a Boolean array, scalar multiplication, or applying math functions, will preserve the index-value relationship.

```python
obj2[obj2 > 0]
```

```python
obj2 * 2
```

```python
np.exp(obj2)
```

Another way to think about a Series is as a **fixed-length, ordered dictionary**, as it is a mapping of index values to data values. It can be used in many context where you might use a dictionary.

```python
'b' in obj2
```

```python
'e' in obj2
```

A series can be created from a dictionary.

```python
sdata = {'Ohio': 35000, 'Texas': 71000, 'Oregon': 16000, 'Utah': 5000}
obj3 = pd.Series(sdata)
obj3
```

A series can be converted back to a dictionary with its to_dict method.

```python
obj3.to_dict()
```

```python
states = ['California', 'Utah', 'Oregon', 'Texas']
obj4 = pd.Series(sdata, index=states)
```

```python
obj4
```

The isna and notna functions in pandas should be used to detect missing data.

```python
pd.isna(obj4)
```

```python
obj4.isna()
```

A useful Series feature for many applications is that it automatically aligns by index label in arithmetic operations:

```python
obj3 + obj4
```

### DataFrame

There are many ways to construct a DataFrame, though one of the most common is from a dictionary of equal-length lists or NumPy arrays.

```python
data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada', 'Nevada'],
        'year': [2000, 2001, 2002, 2001, 2002, 2003],
        'pop': [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}
frame = pd.DataFrame(data)
```

```python
frame
```

```python
# head method selects only the first five rows
frame.head()
```

```python
# tail method selects only the last five rows
frame.tail()
```

If you sepcify a sequence of columns, the DataFrame's columns will be arranged in that order.

```python
pd.DataFrame(data, columns=['year', 'state', 'pop'])
```

If you pass a column that isn't contained in the dictionary, it will appear with missing values in the result:

```python
frame2 = pd.DataFrame(data, columns=['year', 'state', 'pop', 'debt'])
frame2
```

A column in a DataFrame can be retrieved as a Series either by dictionary-like notation or by using the dot attribute notation.

```python
frame2['state']
```

```python
frame2.state
```

Rows can be retrieved by position or name with the special iloc and loc attributes.

```python
frame2.loc[1]
```

```python
frame2.iloc[2]
```

Columns can be modified by assignment. A column can be assigned a scalar value or an array of values.

```python
frame2['debt'] = 16.5
frame2
```

```python
frame2.debt = np.arange(0.0, 6)
frame2
```
