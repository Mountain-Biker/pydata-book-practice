## 5.1 Introduction to pandas Data Structures

### Series

A Series is a one-dimensional array-like object containing a sequence of values of the same type and an associated array of data labels, called its index.

```python
obj = pd.Series([4, 5, -7, 3])
```

```python
obj
```

```python
obj2 = pd.Series([1, 2, 3, 4], index=['a', 'b', 'c', 'd'])
```

```python
obj2
```

```python
obj2.index
```

```python
obj2['a']
```

```python
obj2['b'] = 6
```

```python
obj2[['a', 'b']] = -1
```

```python
obj2
```

Using Numpy functions or Numpy-like operations, such as filtering with a Boolean array, scalar multiplication, or applying math functions, will preserve the index-value relationship.

```python
obj2[obj2 > 0]
```

```python
obj2 * 2
```

```python
np.exp(obj2)
```

Another way to think about a Series is as a **fixed-length, ordered dictionary**, as it is a mapping of index values to data values. It can be used in many context where you might use a dictionary.

```python
'b' in obj2
```

```python
'e' in obj2
```

A series can be created from a dictionary.

```python
sdata = {'Ohio': 35000, 'Texas': 71000, 'Oregon': 16000, 'Utah': 5000}
obj3 = pd.Series(sdata)
obj3
```

A series can be converted back to a dictionary with its to_dict method.

```python
obj3.to_dict()
```

```python
states = ['California', 'Utah', 'Oregon', 'Texas']
obj4 = pd.Series(sdata, index=states)
```

```python
obj4
```

The isna and notna functions in pandas should be used to detect missing data.

```python
pd.isna(obj4)
```

```python
obj4.isna()
```

A useful Series feature for many applications is that it automatically aligns by index label in arithmetic operations:

```python
obj3 + obj4
```

### DataFrame

There are many ways to construct a DataFrame, though one of the most common is from a dictionary of equal-length lists or NumPy arrays.

```python
data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada', 'Nevada'],
        'year': [2000, 2001, 2002, 2001, 2002, 2003],
        'pop': [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}
frame = pd.DataFrame(data)
```

```python
frame
```

```python
# head method selects only the first five rows
frame.head()
```

```python
# tail method selects only the last five rows
frame.tail()
```

If you sepcify a sequence of columns, the DataFrame's columns will be arranged in that order.

```python
pd.DataFrame(data, columns=['year', 'state', 'pop'])
```

If you pass a column that isn't contained in the dictionary, it will appear with missing values in the result:

```python
frame2 = pd.DataFrame(data, columns=['year', 'state', 'pop', 'debt'])
frame2
```

A column in a DataFrame can be retrieved as a Series either by dictionary-like notation or by using the dot attribute notation.

```python
frame2['state']
```

```python
frame2.state
```

Rows can be retrieved by position or name with the special iloc and loc attributes.

```python
frame2.loc[1]
```

```python
frame2.iloc[2]
```

Columns can be modified by assignment. A column can be assigned a scalar value or an array of values.

```python
frame2['debt'] = 16.5
frame2
```

```python
frame2.debt = np.arange(0.0, 6)
frame2
```

When you are assigning lists or arrays to a column, the value's length must match the length of the DataFrame. If you assign a Series, its labels will be realigned exactly to the DataFrame's index, inserting missing values in any index values not present.

```python
val = pd.Series([-1.2, -1.5, -1.7, 1.23], index=['two', 'four', 'five', 0])
```

```python
frame2['debt'] = val
frame2
```

Assigning a column that doesn't exist will create a new column. New column cannot be created with the frame2.eastern dot attribute notation.

```python
frame2['eastern'] = frame2['state'] == 'Ohio'
frame2
```

The `del` method can be used to remove a column.

```python
del frame2['eastern']
```

```python
frame2.columns
```

```python
df1 =  frame2[['year', 'debt']]
print(df1._is_view)

df2 = frame2[['year', 'state']]
print(df2._is_view)
```

## 5.2 Essential Functionality

### Reindexing

```python
obj = pd.Series([4.5, 7.2, -5.3, 3.6], index=['d', 'b', 'a', 'c'])
```

```python
obj
```

Calling reindex on the Series rearranges the data according to the new index, introducing missing values if any index values were not already present.

```python
obj2 = obj.reindex(['a', 'b', 'c', 'd', 'e'])
```

```python
obj2
```

```python
obj3 = pd.Series(['blue', 'purple', 'yellow'], index=[0, 2, 4])
```

```python
obj3
```

```python
obj3.reindex(np.arange(6), method='ffill')
```

With DataFrame, reindex can alter the row index, columns or both.

```python
frame = pd.DataFrame(np.arange(9).reshape((3, 3)),
                     index=['a', 'c', 'd'],
                     columns=['Ohio', 'Texas', 'California'])
```

```python
frame
```

```python
frame2 = frame.reindex(index=['a', 'b', 'c', 'd'])
frame2
```

The columns can be reindexed with the columns keyword. Because 'Ohio' was not in states, the data for that column is dropped from the result.

```python
states = ['Texas', 'Utah', 'California']
frame3 = frame.reindex(columns=states)
frame3
```

We can also reindex by using the loc operator, and many users prefer to do it this way. This works only if all of the new index labels already exist in the DataFrame (whereas reindex will insert missing data for new labels)

```python
frame
```

```python
frame4 = frame.loc[['a', 'd', 'c'], ['Texas', 'Ohio']]
frame4
```

```python
frame4._is_view
```

### Pitfalls with Chained indexing

When trying to modify values in a dataframe in place, we should always use loc and iloc.

```python
data.loc[data.three > 5]['three'] = 6
```

The above code will return a SettingWithCopyWarning, because `data.loc[data.three > 5]` will return a copy of the original DataFrame. So the code will modify the copy of the dataframe but not the origial one. To fix the issue, we should use a single loc operation:

```python
data.loc[data.three > 5, 'three'] = 6
```

```python
data
```

### Arithmetic and Data Alignment

When you add objects, if any index pairs are not the same, the respective index in the result will be the union of the index pairs. The internal data alignment introduces missing values in the label locations that don't overlap.

```python
s1 = pd.Series([7, -2, 3, 1], index=['a', 'b', 'c', 'e'])
```

```python
s2 = pd.Series([2, 3, 4], index=['a', 'c', 'd'])
```

```python
s1 + s2
```

```python
df1 = pd.DataFrame(np.arange(0, 9.0).reshape((3, 3)),
                   columns=list('bcd'),
                   index=['Utah', 'Colorado', 'Nevada'])
df1
```

```python
df2 = pd.DataFrame(np.arange(0, 12.0).reshape((4, 3)),
                   columns=list('bde'),
                   index=['Ohio', 'Utah', 'Nevada', 'Colorado'])
df2
```

```python
df1 + df2
```
